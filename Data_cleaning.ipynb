```python
#importing the necessary libraries :
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel(r"C:\Users\ABWYD\Downloads\Customer Call List.xlsx")
df
```

```python
# Creating a cope from the data to clean it : 
# importing the library copy
import copy 
```

```python
df1= copy.deepcopy(df)
```

```python
# 1 - checking if there are duplicates in the data : 
df1[df1.duplicated()]
# 1 data found: custoerId (20)
# removing the duplicated data : 
df1.drop_duplicates(inplace=True)
```

```python
#2- removing any unnecessary columns : 
#  we have 1 unneeded column (Not_Useful_Column)
df1.drop(columns='Not_Useful_Column',inplace=True)
```

```python
#3- checking if the fisrt name column is clean:
df1[df1['First_Name'].str.len() != df1['First_Name'].str.strip().str.len()]
# 1 row found , index(16):
df1.loc[16,'First_Name']
# # space found in the end of the name: 
df1.loc[16,'First_Name'] = df1.loc[16,'First_Name'].strip()

```

```python
# 4 cleaning the last name column: 
df1['Last_Name']
# data is not clean, (/.._) found:
df1['Last_Name']= df1['Last_Name'].str.strip('/._ ',)
```

```python
# 5- cleaning the phone number column: 
df1['Phone_Number']
# the data is not formated properly, and not in the same format 
# first, removing all the (/-| ):
df1['Phone_Number']=df1['Phone_Number'].str.replace('[^A-Za-z0-9]','',regex=True)    # this code will romve any thing that is not charecter or digit
```

```python
# formating the numbers into united format: 
# first changing the data type into str in order to format it :
df1['Phone_Number']= df1['Phone_Number'].apply(lambda x: str(x))
```

```python
# now that it is transported into sre type, we format the data: 
# the format will be as : "123-456-7891"

### WORNING, DO NOT RUN THIS CODE MORE THAN ONCE, OTHERWISE IT WILL RUIN THE DATA : ###
df1['Phone_Number']= df1['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10]) 
```

```python
# romoving the empty data: 
df1['Phone_Number']=df1['Phone_Number'].str.replace('nan--','')
df1['Phone_Number']=df1['Phone_Number'].str.replace('Na--','')
```

```python
# 6 cleaning the Address column by spliting the data into (street, Satate, zip_code)
df1[['Street','State','Zip_Code']]=df1['Address'].str.split(',',expand=True)
df1
```

```python
# droping the address column: 
df1.drop(columns = 'Address',inplace = True)
```

```python
# 7 formating the Paying Customer,Do_Not_Contact columns : 
df1['Paying Customer']= df1['Paying Customer'].replace({'Y':'Yes','N':'No'})
df1['Do_Not_Contact'] = df1['Do_Not_Contact'].replace({'Y':'Yes','N':'No'})
```

```python
# filling the empty data in the Do_Not_Contact column:
df1['Do_Not_Contact']=df1['Do_Not_Contact'].fillna('No')
```

```python
# Droping the data if the customer specified ( Do_Not_Contact = Yes):
# first, stripping the column:
df1['Do_Not_Contact']=df1['Do_Not_Contact'].str.strip()
df1 = df1[df1['Do_Not_Contact'] != 'Yes']
```

```python
# droping data if the phone number not provided:
df1 = df1[df1['Phone_Number']!='']
```

```python
# i found that all the data in the zip_code column is not provided, then the colum is not needed :
df1= df1.drop(columns= 'Zip_Code')
```

```python
# final step, resetting the index:
df1.reset_index(inplace=True)
df1=df1.drop(columns= 'index')
```
